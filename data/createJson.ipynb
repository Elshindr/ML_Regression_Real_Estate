```python
import pandas as pd
from faker import Faker
import numpy as np
import json
import random
from datetime import datetime

def valeur_ou_manquante(valeur, proba_manquant=0.08):
    return None if random.random() < proba_manquant else valeur

city = ["Paris", "Lyon", "Marseille", "Toulouse", "Bordeaux", "Lille", "Nantes", "Nice"]
type = ["apartment", "house", "studio"]
datas = [{
    "city": "Paris",
    "property_type": "apartment",
    "surface": 65.5,
    "rooms": 3,
    "bedrooms": 2,
    "bathrooms": 1,
    "price": 420000,
    "year_built": 1998,
}]
fake = Faker()


for i in range(1000):

    surface = fake.pyfloat(min_value=23, max_value=120)
    rooms = fake.random_int(min=2, max=13)
    bedrooms = fake.random_int(min=2, max=rooms)
    bathrooms = fake.random_int(min=2, max=rooms)
    price=fake.random_int(min=10000, max=100000)
    year=fake.random_int(min=1980, max=2023)
    
    if random.random() < 0.1:
        surface = round(random.choice([-199.9, 250.0, 1999.0]), 1)
    if random.random() < 0.1:
        price = random.choice([-1, 150000, 199999])
    if random.random() < 0.1:
        bathrooms = random.choice([-5, 9999])

    obj = {
        "city": valeur_ou_manquante(fake.random_element(elements=city)),
        "property_type": valeur_ou_manquante(fake.random_element(elements=type)),
        "surface": valeur_ou_manquante(surface),
        "rooms": valeur_ou_manquante(rooms),
        "bedrooms": valeur_ou_manquante(bedrooms),
        "bathrooms": bathrooms,
        "price": price,
        "year_built": year
    }
    datas.append(obj)

# --- doublons ---
nb_duplicats = random.randint(12,25)
duplicats = [dict(random.choice(datas)) for _ in range(nb_duplicats)]
datas.extend(duplicats)
random.shuffle(datas)

print(f"Total lignes : {len(datas)} (dont {nb_duplicats} doublons)")


json_str = json.dumps(datas, indent=4)
with open("price_raw.json", "w") as f:
    f.write(json_str)
```
